#### Level 1: Data Loading 

```python
import pandas as pd
df = pd.read_csv('data/SuperStore_Sales_Dataset.csv')
df.index
```

```python
# How many rows and columns does the dataset contain?
print("Shape:", df.shape)

# What are the column names and their data types?
print(df.columns)
df.info()

# Preview the first and last five records
display(df.head())
display(df.tail())
```

#### level 2: Data Selection

```python
# Select only the sales-related columns
sales_columns = df[
    ["Order ID",
     "Customer Name",
     "Category",
     "Sales",
     "Profit"]
]

display(sales_columns.head())

# Display only the customer names
display(df["Customer Name"])

# Display the first 10 orders
display(df.iloc[:10])
```

#### level 3: Filtering

```python
# Find all orders with Sales greater than 500
high_sales = df[df["Sales"] > 500]

display(high_sales.head())

# Find all Technology products
technology = df[df["Category"] == "Technology"]

display(technology.head())

# Find all orders with negative Profit
loss_orders = df[df["Profit"] < 0]

display(loss_orders.head())
```

#### Level 4  Basic Cleaning

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d-%m-%Y"
)
```

```python
# Check for missing values
print(df.isnull().sum())

# Check for duplicate rows
print(df.duplicated().sum())

# Convert Order Date and Ship Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

#### Level 5 GroupBy

```python
# What are the total Sales and Profit by Category?
category_metrics = df.groupby("Category")[["Sales", "Profit"]].sum()

display(category_metrics)

# What are the total Sales by Region?
sales_by_region = df.groupby("Region")["Sales"].sum()

display(sales_by_region)

# What is the average Sales for each Segment?
average_sales_segment = (
    df.groupby("Segment")["Sales"]
      .mean()
)

display(average_sales_segment)
```

#### Level 6 Customer Analysis

```python
# Which customers generated the highest Sales?
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

display(top_customers)

# Which customers generated the highest Profit?
top_profit_customers = (
    df.groupby("Customer Name")["Profit"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

display(top_profit_customers)

# How many orders does each customer have?
customer_orders = (
    df.groupby("Customer Name")["Order ID"]
      .count()
      .sort_values(ascending=False)
)

display(customer_orders.head(10))
```

#### Level 7 Region Analysis

```python
# Which region has the highest Sales and Profit?
region_metrics = (
    df.groupby("Region")[["Sales", "Profit"]]
      .sum()
)

display(region_metrics)

# Which region has the lowest Sales?
lowest_region = (
    df.groupby("Region")["Sales"]
      .sum()
      .sort_values()
)

display(lowest_region)

# How many orders came from each Region?
orders_by_region = (
    df.groupby("Region")["Order ID"]
      .count()
)

display(orders_by_region)
```

#### Level 8 Category Analysis

```python
# Which Category generated the highest Sales?
sales_category = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

display(sales_category)

# Which Sub-Category generated the highest Profit?
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

display(subcategory_profit.head(10))

# Which products generated the highest Sales?
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

display(top_products)
```

#### Level 9 Shipping & Payment

```python
# How many orders used each Ship Mode?
ship_mode = (
    df.groupby("Ship Mode")["Order ID"]
      .count()
)

display(ship_mode)

# How many orders used each Payment Mode?
payment_mode = (
    df.groupby("Payment Mode")["Order ID"]
      .count()
)

display(payment_mode)

# What is the average shipping time for each Ship Mode?
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

average_shipping = (
    df.groupby("Ship Mode")["Shipping Days"]
      .mean()
)

display(average_shipping)
```

####  Level 10  Time Analysis

```python
# How did Sales and Profit vary by Year?
yearly_metrics = (
    df.groupby(df["Order Date"].dt.year)[["Sales", "Profit"]]
      .sum()
)

display(yearly_metrics)

# How did Sales and Profit vary by Month?
monthly_metrics = (
    df.groupby(df["Order Date"].dt.month_name())[["Sales", "Profit"]]
      .sum()
)

display(monthly_metrics)

# How has shipping volume changed over time?
shipping_by_year = (
    df.groupby(df["Ship Date"].dt.year)
      .size()
)

display(shipping_by_year)

# Additional time-series used in Plots notebook
shipping_by_month = (
    df.groupby(df["Ship Date"].dt.month_name())
      .size()
)

shipping_by_date = (
    df.groupby(df["Ship Date"].dt.date)
      .size()
      .sort_index()
)
```
