
### Converting messy text dates into real datetime objects, extract months using .dt, group data by month, and visualize monthly shipping trends with a bar chart.

```python
import pandas as pd
df = pd.read_csv('data/SuperStore_Sales_Dataset.csv')
```

```python
df.columns
```

```python
df["Ship Date"].sample(20)
```

```python
## to make the data consistent, we will convert the "Ship Date" column to datetime format using the `pd.to_datetime()` function. This will help us analyze the shipping trends more effectively.
df2 = df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="mixed",
    dayfirst=True
)
```

```python
df2
```

```python
shipping_by_year = df.groupby(
    df["Ship Date"].dt.year

).size()
```

```python
print(shipping_by_year)
shipping_by_year.plot(kind="bar", title="Shipping by Year", figsize=(5, 5))
```
